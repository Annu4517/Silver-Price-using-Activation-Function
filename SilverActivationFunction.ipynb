```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


df = pd.read_csv("silver_price.csv")


numeric_cols = df.select_dtypes(include=[np.number]).columns
x = df[numeric_cols[0]].values



def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / np.sum(exp_x)



plt.figure()
plt.plot(x, sigmoid(x))
plt.title("Sigmoid Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.show()
plt.close()


plt.figure()
plt.plot(x, tanh(x))
plt.title("Tanh Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.show()
plt.close()



plt.figure()
plt.plot(x, relu(x))
plt.title("ReLU Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.show()
plt.close()



plt.figure()
plt.plot(x, leaky_relu(x))
plt.title("Leaky ReLU Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.show()
plt.close()


x_small = x[:20]
x_small_norm = (x_small - np.mean(x_small)) / np.std(x_small)

y_softmax = softmax(x_small_norm)


plt.figure(figsize=(8, 4))
plt.plot(range(len(y_softmax)), y_softmax, marker='o')
plt.title("Softmax Activation on Silver Price (Line Plot)")
plt.xlabel("Index")
plt.ylabel("Probability")
plt.grid(True)
plt.tight_layout()
plt.show()
```
